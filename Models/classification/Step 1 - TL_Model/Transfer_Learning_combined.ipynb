This function processes a dictionary of patient data frames (`datadict`), dropping
unnecessary columns and converting the remaining data to NumPy arrays for efficient
numerical operations.

Args:
    datadict (dict): A dictionary where keys are patient IDs and values are pandas DataFrames
        containing patient data.
    bool_drop_ABS (bool, optional): A flag indicating whether to drop additional columns
        beyond the default set. Defaults to False.

Returns:
    None

Prints:
    - The number of patients in the dataset.
    - The shape of the processed data frame for each patient.
    - A sample of the head (first 5 rows) of the processed data frame for a single patient.

Raises:
    ValueError: If `bool_drop_ABS` is True and any columns other than `id`, `date`,
                 and `am_pef_org` are present in the data frames, as they might not be
                 convertible to NumPy float32.

````python
def process_patient_data(datadict, bool_drop_ABS=False):
    p_ids = datadict.keys()
    sample_size_dict = {}
    print("patients: n =", len(p_ids), end="\n\n")

    for p_id in p_ids:
        df = datadict[p_id]

        # Drop common unnecessary columns
        df.drop(["id", "date", "am_pef_org"], axis=1, inplace=True)

        # Optionally drop additional columns for further processing
        if bool_drop_ABS:
            valid_cols = set(df.columns) - {"id", "date", "am_pef_org"}
            if valid_cols:
                raise ValueError(
                    "Unexpected columns found with bool_drop_ABS=True: {}".format(
                        valid_cols
                    )
                )
            # No additional columns found, proceed with conversion

        # Convert data frame to NumPy array and cast to float32
        df = np.asarray(df).astype(np.float32)
        print(p_id, "shape:", df.shape)
        sample_size_dict[p_id] = df.shape[0]

    return sample_size_dict, p_ids, df


sample_size_dict, p_ids, df = process_patient_data(datadict, bool_drop_ABS=True)
````

## Function: `split_data`

### Description:
This function splits a given dataset into training and testing subsets according to the specified ratio.

### Arguments:
- `data` (DataFrame): The dataset to be split.
- `test_size` (float): The proportion of the dataset to include in the test split. Should be between 0.0 and 1.0.
- `random_state` (int, optional): Controls the randomness of the sample. Defaults to None.

### Returns:
tuple: A tuple containing the training dataset and the testing dataset, in that order.

### Examples:
```python
train_data, test_data = split_data(data, test_size=0.2, random_state=42)


````python
def split_data(
    target, data, bool_remove_target, transfer=False
) -> (pd.DataFrame, pd.DataFrame):
    """
    Creates a combined data set from all the ids in the given dictionary,
    If transfer==false, returns just target df
    If transfer==True and bool_remove_target==True, returns all but the target df.
    If transfer==True and bool_remove_target==False, return all patients’ data

    :param target: Who the target is for prediction.
    :param data: Dict of DataFrames for prediction.
    :param bool_remove_target: Boolean flag indicating whether to remove the target DataFrame or not.
    :param transfer: Boolean flag indicating whether to transfer data.
    :return X: Predictive labels.
    :return Y: Dependent label.
    """
    # Create an empty DataFrame to hold the combined data
    dataset = pd.DataFrame()

    # Check if data transfer is requested
    if transfer == True:
        # If transfer is requested and removal of target DataFrame is also requested
        if bool_remove_target == True:
            # Concatenate all DataFrames except the target DataFrame
            dataset = pd.concat(
                [data[key].copy() for key in data.keys() if key != target]
            )
        else:
            # Concatenate all DataFrames including the target DataFrame
            dataset = pd.concat([data[key].copy() for key in data.keys()])

    else:
        # If transfer is not requested, simply copy the target DataFrame
        dataset = data[target].copy()

    # Extract features (X) by dropping the target variable
    X = dataset.drop([CLASS_VAR], axis=1)
    # Extract the target variable (Y)
    Y = dataset.loc[:, CLASS_VAR:CLASS_VAR]

    return X, Y
````

## Function: `data_augmentation`

### Description:
This function is used to augment a given DataFrame using a specified technique to balance the dataset, particularly when dealing with imbalanced classes. It allows for plugging in different augmentation algorithms by providing the appropriate function call and ensuring compatibility with the input and output formats.

### Arguments:
- `augmentation_algo` (str): Name of the augmentation technique to be applied.
- `train_dataset` (DataFrame): DataFrame representing the dataset to be augmented, containing both feature variables (X) and labels (Y). Certain columns are dropped: ["id", "date", "am_pef_org", "BMI", "sex", "age"].
- `classIndex` (str): Label of the class with minority/majority split.
- `minorityLabel` (str): The value of `classIndex` representing the minority class.
- `printDebug` (bool, optional): Flag for enabling debug printing. Defaults to False.

### Returns:
DataFrame: Augmented dataset, similar to `train_dataset` but with synthetic rows included to achieve balance.

### Notes:
- Other augmentation algorithms can be plugged in by ensuring compatibility with the function signature and return type.
- Different augmentation techniques are implemented based on the value of `augmentation_algo`.
- Upon successful augmentation, the function prints out the value counts of the class labels to ensure balance.
- If the specified augmentation technique is not implemented, a `NotImplementedError` is raised.

### Examples:
```python
augmented_df = data_augmentation("ADKNN", train_df, "label", "minority", printDebug=True)

````python
def data_augmentation(
    augmentation_algo,
    train_dataset,
    classIndex,
    minorityLabel,
    printDebug=False,
):
    """
    Augments the given dataframe via the specified technique.

    Args:
        augmentation_algo (str): Name of the augmentation technique to be used.
        train_dataset (DataFrame): DataFrame to be augmented, containing both feature variables (X) and labels (Y).
                                   It represents the complete dataset except the train split.
                                   Certain columns are dropped: ["id", "date", "am_pef_org", "BMI", "sex", "age"].
        classIndex (str): Label of the class with minority/majority split.
        minorityLabel (str): The value of classIndex representing the minority class.
        printDebug (bool, optional): Flag for enabling debug printing. Defaults to False.

    Returns:
        DataFrame: Augmented dataset, similar to train_dataset but with synthetic rows included to achieve balance.

    Note:
        - Other augmentation algorithms can be plugged in by ensuring compatibility with the function signature and return type.
        - For the augmentation technique, 'ADKNN', the `balance_method.ADKNN` function is used.
        - For the augmentation technique, 'AutoEncoder_1', the `balance_method.generate_autoencoder` function is used.
        - For the augmentation technique, 'GammaSmoteBoost', the `balance_method.GammaSMOTEBoostCC` function is used.
        - For the augmentation technique, 'SDDSMOTEBoostCC', the `balance_method.SDDSMOTEBoostCC` function is used.
        - For the augmentation technique, 'SMOTEBoostCC', the `balance_method.SMOTEBoostCC` function is used.

    Raises:
        NotImplementedError: If augmentation_algo is not recognized.

    Examples:
        >>> augmented_df = data_augmentation("ADKNN", train_df, "label", "minority", printDebug=True)
    """
    if augmentation_algo == "ADKNN":
        train_dataset = balance_method.ADKNN(
            train_dataset,
            classIndex=classIndex,
            minorityLabel=minorityLabel,
            printDebug=printDebug,
        )
    elif augmentation_algo == "AutoEncoder_1":
        train_dataset = balance_method.generate_autoencoder(
            train_dataset,
            class_var=classIndex,
            minority_var=minorityLabel,
            printDebug=printDebug,
        )
    elif augmentation_algo == "GammaSmoteBoost":
        train_dataset = balance_method.GammaSMOTEBoostCC(
            train_dataset,
            numIterations=5,
            printDebug=printDebug,
        )
    elif augmentation_algo == "SDDSMOTEBoostCC":
        train_dataset = balance_method.SDDSMOTEBoostCC(
            train_dataset,
            printDebug=printDebug,
        )
    elif augmentation_algo == "SMOTEBoostCC":
        train_dataset = balance_method.SMOTEBoostCC(
            train_dataset,
            numIterations=5,
            printDebug=printDebug,
        )
        print(f"augmented data size: {train_dataset.shape}")
    else:
        raise NotImplementedError(
            f"Augmentation technique '{augmentation_algo}' not implemented."
        )

    print("~~~~~~~ value_counts: ~~~~~~~")
    print("SHOULD BE BALANCED NOW:\n", train_dataset[classIndex].value_counts())
    return train_dataset
````

## Function: `population_model`

### Description:
Trains a population-based source model.

### Arguments:
- `pop_target_grp` (list): List of targets to train for, e.g. ["SB-070"]
- `POPULATION_EPOCH_SIZE_LIST` (list): List of Number of epochs for NN
- `num_pop_folds` (int): Number of k-fold cross validations
- `bool_remove_target` (bool): Whether or not to remove target pt from source model training
- `augmentation_algo` (str): Name of the augmentation technique to be applied.

### Returns:
None, but save models to directory. P.S. Needs filepath names resolved above.

### Notes:
- Initializes the sample size.
- Loops through each target group and performs the following steps:
  1. Measures runtime.
  2. Defines a directory for the current patient.
  3. Splits data based on the target group.
  4. Performs k-fold cross-validation.
  5. Iterates over different epoch sizes for neural network training.
  6. Trains neural network, labels, and saves the model.
  7. Appends evaluation results to a dataframe.
  8. Writes evaluation results to a CSV file.

### Examples:
```python
population_model(
    pop_target_grp=["SB-070"],
    POPULATION_EPOCH_SIZE_LIST=[10, 20, 30],
    num_pop_folds=5,
    bool_remove_target=True,
    augmentation_algo="ADKNN"
)

````python
def population_model(
    pop_target_grp,
    POPULATION_EPOCH_SIZE_LIST,
    # PATIENT_EPOCH_SIZE,
    num_pop_folds,
    bool_remove_target,
    augmentation_algo,
):
    """
    Trains a population-based source model.

    :param pop_target_grp: List of targets to train for, e.g. ["SB-070"]
    :param POPULATION_EPOCH_SIZE_LIST: List of Number of epochs for NN
    # :param PATIENT_EPOCH_SIZE: Used to label model for use later
    :param num_pop_folds: Number of k-fold cross validations
    :param bool_remove_target: Whether or not to remove target pt from source model training
    :return: None, but save models to directory. P.S. Needs filepath names resolved above.
    """

    sample_size = ""  # Initialize sample size

    # Loop through each target group
    for i in range(0, len(pop_target_grp)):
        target = pop_target_grp[i]  # Get current target group
        runtime()  # Measure runtime

        # Define directory for current patient
        patient_dir = Path(
            f"{RESULT_DIR}/{Architecture_GROUP}/diagrams_n_population_models/{target}"
        )

        # Create directory for this patient if it doesn't exist
        if not (patient_dir.exists() and patient_dir.is_dir()):
            os.makedirs(patient_dir)

        # Split data based on target group
        X, Y = split_data(
            target=target,
            data=datadict,
            bool_remove_target=bool_remove_target,
            transfer=True,
        )

        # Perform k-fold cross validation
        if num_pop_folds > 1:
            kf = KFold(n_splits=num_pop_folds, shuffle=False).split(X.to_numpy())

        for fold in range(1, num_pop_folds + 1):
            # Make train/test sets
            if num_pop_folds == 1:
                X_train, X_test, Y_train, Y_test = train_test_split(
                    X.to_numpy(), Y.to_numpy(), test_size=0.2, shuffle=True
                )
            else:
                train_index, test_index = next(kf)
                X_train, X_test = X.to_numpy()[train_index], X.to_numpy()[test_index]
                Y_train, Y_test = Y.to_numpy()[train_index], Y.to_numpy()[test_index]

            # Dataframe manipulation for training set
            df_X_train = pd.DataFrame(X_train, columns=X.columns.values)
            df_Y_train = pd.DataFrame(Y_train, columns=Y.columns.values)
            joined_train_dataset = df_X_train.join(df_Y_train)

            # Augment data
            train_dataset = data_augmentation(
                augmentation_algo,
                joined_train_dataset,
                classIndex=CLASS_VAR,
                minorityLabel=minority_variable,
            )

            # Convert augmented data back to numpy arrays
            X_train = train_dataset.drop(CLASS_VAR, axis=1).to_numpy()
            Y_train = train_dataset.loc[:, CLASS_VAR:CLASS_VAR].to_numpy()

            # Initialize dataframe for storing evaluation results
            source_evaluation_results = pd.DataFrame(columns=EVALUATION_METRICS)

            # Iterate over different epoch sizes for neural network training
            for POPULATION_EPOCH_SIZE in POPULATION_EPOCH_SIZE_LIST:
                print(
                    "~~~~~~~~~~~",
                    POPULATION_EPOCH_SIZE,
                    "~~~~~~~~~~~~~~~~~~~",
                    target,
                    "~~~~~~~~~~~~~~~~~~~",
                )

                # Train neural network
                print(f"Population Fold {fold}: Training data size = {len(X_train)}")

                # Initialize and train neural network
                net = network_tl(
                    X_train, Y_train, n_hidden=32, learning_rate=0.001, device=device
                )
                net.set_Y_tind(Y_train)
                net.make_pop(Architecture_GROUP)
                net.train(n_iters=POPULATION_EPOCH_SIZE)

                # Update sample size if target removal is not required
                if bool_remove_target == False:
                    sample_size = f"{str(len(X_train))}"

                # Label and save model
                VERSION_TAG = f"{target}_size_{sample_size}_{Architecture_GROUP}_{POPULATION_EPOCH_SIZE}_POP_FOLD_{fold}"
                modelname = f"{VERSION_TAG}.keras"
                training_curve_file = patient_dir / f"{VERSION_TAG}_training_curve"

                model_path = patient_dir / modelname
                net.nn_pop.save(model_path)  # nn_pop
                net.save_training_curve(training_curve_file)
                source_results, source_c_matrix = net.pop_test(X_test, Y_test)

                # Append evaluation results to dataframe
                source_evaluation_results.loc[f"{VERSION_TAG}"] = list(source_results)

            # Define output path for evaluation metrics
            if bool_remove_target == True:
                output_path = os.path.join(
                    f"{RESULT_DIR}/{Architecture_GROUP}/diagrams_n_population_models",
                    f"Pop_{Architecture_GROUP}_evaluation_metrics_TargetRemoved.csv",
                )
            else:
                output_path = os.path.join(
                    f"{RESULT_DIR}/{Architecture_GROUP}/diagrams_n_population_models",
                    f"Pop_{Architecture_GROUP}_evaluation_metrics.csv",
                )

            # Write evaluation results to CSV file
            if os.path.isfile(output_path):
                source_evaluation_results.to_csv(output_path, mode="a", header=False)
            else:
                source_evaluation_results.to_csv(output_path, mode="a")

    print("evaluation results saved")
````

- `POPULATION_EPOCH_SIZE_LIST` (list): List of Number of epochs for NN.

### Returns:
list: List containing names of population models found in the target directory.

### Notes:
- Initializes a list to store population model names.
- Defines the path for population models based on the provided parameters.
- Searches for population models based on the provided target ID and epoch size list.
- Returns a list of population model names found in the target directory.

### Example:
```python
pop_model_list = find_pop_model(
    target_id="SB-070",
    RESULT_DIR="/path/to/results",
    Architecture_GROUP="arch_group",
    POPULATION_EPOCH_SIZE_LIST=[10, 20, 30]
)
print(pop_model_list)

````python
def find_pop_model(
    target_id, RESULT_DIR, Architecture_GROUP, POPULATION_EPOCH_SIZE_LIST
):
    """
    Returns all population model names found in target directory
    :param target_id: Participant ID of models to search for
    :param RESULT_DIR: Results directory
    :param Architecture_GROUP: Sub directory of results to search in.
    """
    pop_model_list = []  # Initialize list to store population model names
    pop_model_path = f"{str(RESULT_DIR)}/{Architecture_GROUP}/diagrams_n_population_models/{target_id}"

    # Find population model based on provided parameters
    for POPULATION_EPOCH_SIZE in POPULATION_EPOCH_SIZE_LIST:
        pop_model_list.extend(
            file_name
            for file_name in os.listdir(pop_model_path)
            if (
                file_name[-6:] == ".keras"
                and "_" + str(POPULATION_EPOCH_SIZE) + "_" in file_name
                and target_id in file_name
            )
        )
    return pop_model_list


# If in testing mode, execute the following block of code
if bool_testing:
    target_id = pop_target_grp[0]  # Get the target ID for testing
    RESULT_DIR = Path(
        f"{execution_path}/Output_Results/{output_folder}/"
    )  # Define result directory
    # Find population model file list
    pop_model_file_list = find_pop_model(
        target_id, RESULT_DIR, Architecture_GROUP, POPULATION_EPOCH_SIZE_LIST
    )
    display(pop_model_file_list)


@tf.function(reduce_retracing=True)
def predict_with_model(model, input_data):
    """
    Decorating with a @tf.function as this function is to be called inside a for loop
    """
    return model(input_data)  # Predict with the provided model and input data
````

## Function: `individual_model`

### Description:
Transfer learning for an individual target. Takes a previously trained source model and refines it for `target_id`.

### Arguments:
- `target_id` (str): Participant ID to refine the model on.
- `RESULT_DIR` (str): Directory where to save results.
- `Architecture_GROUP` (str): Type of architecture to use (string of how many layers are unfrozen).
- `POPULATION_EPOCH_SIZE_LIST` (list): List of epochs for the base model.
- `PATIENT_EPOCH_SIZE_LIST` (list): List of epochs to refine the model with.
- `datadict` (dict): Dictionary of data to use.
- `bool_remove_target` (bool): Indicates whether to remove the target or not.
- `EVALUATION_METRICS` (list): List of evaluation metrics to use.
- `bool_remove_ind_cv` (bool): Indicates whether to remove individual cross-validation or not.
- `num_ind_folds` (int, optional): Number of k-folds for refinement. Defaults to 3.

### Returns:
None.

### Notes:
- Adjusts `num_ind_folds` if necessary.
- Measures runtime.
- Defines the patient directory.
- Splits data based on the target ID.
- Performs k-fold cross-validation.
- Initializes and trains the neural network.
- Retrains every model on `target_id`.
- Saves the model and retraining curve.
- Calculates evaluation metrics and saves them to a CSV file.

### Example:
```python
individual_model(
    target_id="SB-070",
    RESULT_DIR="/path/to/results",
    Architecture_GROUP="arch_group",
    POPULATION_EPOCH_SIZE_LIST=[10, 20, 30],
    PATIENT_EPOCH_SIZE_LIST=[5, 10],
    datadict=datadict,
    bool_remove_target=True,
    EVALUATION_METRICS=["accuracy", "precision", "recall"],
    bool_remove_ind_cv=False,
    num_ind_folds=3
)

````python
def individual_model(
    target_id,
    RESULT_DIR,
    Architecture_GROUP,
    POPULATION_EPOCH_SIZE_LIST,
    PATIENT_EPOCH_SIZE_LIST,
    datadict,
    bool_remove_target,
    EVALUATION_METRICS,
    bool_remove_ind_cv,
    num_ind_folds=3,
):
    """
    Transfer learning for individual target. Takes a previously trained source model and refines it for target_id.

    :param target_id: Participant ID to refine model on
    :param RESULT_DIR: Where to save results
    :param Architecture_GROUP: What type of architecture to use (string of how many layers are unfrozen)
    :param POPULATION_EPOCH_SIZE: Number of epochs of base model
    :param PATIENT_EPOCH_SIZE: Number of epochs to refine model with
    :param datadict: Dictionary of data to use
    :param bool_remove_target: Some confusion here, see targetSet comment below
    :param num_ind_folds: Number of k-folds for refinement
    :param EVALUATION_METRICS: List of evaluation metrics to use
    :returns: Saves model to /diagrams_n_individual_models/{target_id}/
    """

    if bool_remove_ind_cv:
        num_ind_folds = 1
    else:
        if num_ind_folds < 3 or num_ind_folds > 5:
            print(
                f"num_ind_folds can only be set between 3 and 5, but current value is {num_ind_folds}"
            )
            num_ind_folds = 3 if num_ind_folds < 3 else 5
            print(f"num_ind_folds has been updated to {num_ind_folds}")

    runtime()

    # Define patient directory
    patient_dir = Path(
        f"{RESULT_DIR}/{Architecture_GROUP}/diagrams_n_individual_models/{target_id}"
    )
    # Create directory for this patient if it doesn't exist
    if not (patient_dir.exists() and patient_dir.is_dir()):
        os.makedirs(patient_dir)

    # Split data based on target ID
    X, Y = split_data(
        target=target_id,
        data=datadict,
        bool_remove_target=bool_remove_target,
        transfer=False,
    )

    # KFold requires n > 1, need special case for no k-fold
    if num_ind_folds > 1:
        kf = KFold(n_splits=num_ind_folds, shuffle=True).split(X.to_numpy())

    for fold in range(1, num_pop_folds + 1):
        # Make train/test sets
        if num_ind_folds == 1:
            X_train, X_test, Y_train, Y_test = train_test_split(
                X.to_numpy(), Y.to_numpy(), test_size=0.2, shuffle=True, random_state=42
            )
        else:
            train_index, test_index = next(kf)
            X_train, X_test = X.to_numpy()[train_index], X.to_numpy()[test_index]
            Y_train, Y_test = Y.to_numpy()[train_index], Y.to_numpy()[test_index]

        # Create train dataset as DataFrame
        df_X_train = pd.DataFrame(X_train, columns=X.columns.values)
        df_Y_train = pd.DataFrame(Y_train, columns=Y.columns.values)
        train_dataset = df_X_train.join(df_Y_train)

        # Data augmentation
        train_dataset = data_augmentation(
            augmentation_algo,
            train_dataset,
            classIndex=CLASS_VAR,
            minorityLabel=minority_variable,
        )

        target_evaluation_results = pd.DataFrame(columns=EVALUATION_METRICS)
        kfold_confusion_matrix_list = []

        X_train = train_dataset.drop(CLASS_VAR, axis=1).to_numpy()
        Y_train = train_dataset.loc[:, CLASS_VAR:CLASS_VAR].to_numpy()
        print(f"Fold {fold}: Training data size = {len(X_train)}")

        # Initialize and train neural network
        net = network_tl(
            X_train, Y_train, n_hidden=32, learning_rate=0.001, device=device
        )
        net.set_Y_tind(Y_train)

        if bool_remove_target:
            pop_model_file_list = find_pop_model(
                target_id,
                RESULT_DIR,
                Architecture_GROUP,
                POPULATION_EPOCH_SIZE_LIST,
            )
        else:
            pop_model_file_list = find_pop_model(
                "source", RESULT_DIR, Architecture_GROUP, POPULATION_EPOCH_SIZE_LIST
            )

        # Retrain every model on target ID
        for pop_model_file in pop_model_file_list:
            for PATIENT_EPOCH_SIZE in PATIENT_EPOCH_SIZE_LIST:
                print(
                    f" ~~~~~ Patient: {target_id} ~~~~~ Fold: {fold} ~~~~~ Pop_Model_File: {pop_model_file} ~~~~~ PATIENT EPOCH SIZE: {PATIENT_EPOCH_SIZE} ~~~~~"
                )
                if bool_remove_target:
                    pop_model_file_path = f"{str(RESULT_DIR)}/{Architecture_GROUP}/diagrams_n_population_models/{target_id}/{pop_model_file}"
                    VERSION_TAG = (
                        f"{pop_model_file[:-6]}_{PATIENT_EPOCH_SIZE}_IND_FOLD_{fold}"
                    )
                else:
                    pop_model_file_path = f"{str(RESULT_DIR)}/{Architecture_GROUP}/diagrams_n_population_models/source/{pop_model_file}"
                    VERSION_TAG = f"{pop_model_file[:-6]}_{target_id}-{PATIENT_EPOCH_SIZE}_IND_FOLD_{fold}"

                pretrained_model = load_model(
                    pop_model_file_path, compile=False
                )  # This model file contain the model architecture and its weights
                net.make_ind(Architecture_GROUP, pretrained_model)
                net.retrain(X_train, Y_train, n_iters=PATIENT_EPOCH_SIZE)

                predict_with_model(net.nn_ind, X_test)

                # Save model
                ind_model_file_path = f"{str(RESULT_DIR)}/{Architecture_GROUP}/diagrams_n_individual_models/{target_id}/{VERSION_TAG}.keras"
                net.nn_ind.save(ind_model_file_path)

                # Save retraining curve
                training_curve_file = f"{str(RESULT_DIR)}/{Architecture_GROUP}/diagrams_n_individual_models/{target_id}/{VERSION_TAG}_retraining_curve"
                net.save_training_curve(training_curve_file)

                # results for the individual model for this particular fold
                target_results, target_c_matrix = net.test(X_test, Y_test)

                # adding up the results of the confusion matrix for all the folds of the individual model
                kfold_confusion_matrix_list += [target_c_matrix]

                # adding up the results (evaluation metrics) of all the folds of the individual model
                target_evaluation_results.loc[VERSION_TAG] = list(target_results)

        fold += 1

        if bool_remove_target == True:
            output_path = os.path.join(
                f"{RESULT_DIR}/{Architecture_GROUP}/diagrams_n_individual_models",
                f"Ind_{Architecture_GROUP}_evaluation_metrics_TargetRemoved.csv",
            )
        else:
            output_path = os.path.join(
                f"{RESULT_DIR}/{Architecture_GROUP}/diagrams_n_individual_models",
                f"Ind_{Architecture_GROUP}_evaluation_metrics.csv",
            )

        if os.path.isfile(output_path):
            target_evaluation_results.to_csv(output_path, mode="a", header=False)
        else:
            target_evaluation_results.to_csv(output_path, mode="a")

    print("evaluation results saved")
````

````python
def individual_model_wrapper(
    bool_testing: bool,
    bool_execute_ind_model: bool,
    ind_target_grp: List[str],
    RESULT_DIR: str,
    Architecture_GROUP: str,
    POPULATION_EPOCH_SIZE_LIST: List[int],
    PATIENT_EPOCH_SIZE_LIST: List[int],
    datadict: dict,
    bool_remove_target: bool,
    EVALUATION_METRICS: List[str],
    bool_remove_ind_cv: bool,
    num_ind_folds: int,
) -> None:
    """
    Wrapper function for executing individual models.

    This function checks if testing mode is enabled and if individual models
    should be executed. If both conditions are met, it iterates through the
    target groups and calls the individual_model function for each target group.

    Args:
    - bool_testing (bool): Flag indicating whether the code is running in testing mode.
    - bool_execute_ind_model (bool): Flag indicating whether to execute individual models.
    - ind_target_grp (List[str]): List of target groups for individual modeling.
    - RESULT_DIR (str): Directory path to save results.
    - Architecture_GROUP (str): Grouping identifier for architecture selection.
    - POPULATION_EPOCH_SIZE_LIST (List[int]): List of epoch sizes for population modeling.
    - PATIENT_EPOCH_SIZE_LIST (List[int]): List of epoch sizes for patient modeling.
    - datadict (dict): Dictionary containing data for modeling.
    - bool_remove_target (bool): Flag indicating whether to remove the target variable.
    - EVALUATION_METRICS (List[str]): List of evaluation metrics to be used.
    - bool_remove_ind_cv (bool): Flag indicating whether to remove individual cross-validation.
    - num_ind_folds (int): Number of folds for cross-validation in individual modeling.

    Returns:
    None

    """
    if not bool_testing:  # If not in testing mode
        if bool_execute_ind_model:  # If individual model execution is requested
            for target_id in ind_target_grp:  # Iterate through target groups
                individual_model(
                    target_id,
                    RESULT_DIR,
                    Architecture_GROUP,
                    POPULATION_EPOCH_SIZE_LIST,
                    PATIENT_EPOCH_SIZE_LIST,
                    datadict,
                    bool_remove_target,
                    EVALUATION_METRICS,
                    bool_remove_ind_cv,
                    num_ind_folds,
                )
                print("Done with ", target_id)
````

## Function: `segregate_identifier`

### Description:
This function segregates identifiers based on the specified column return type. It iterates through different types of identifiers and returns the corresponding value based on the specified column return type.

### Arguments:
- `index` (str): Identifier value.
- `column_return_type` (str): Type of column to return.
- `POPULATION_EPOCH_SIZE_LIST` (List[int]): List of epoch sizes for population modeling.
- `PATIENT_EPOCH_SIZE_LIST` (List[int]): List of epoch sizes for patient modeling.
- `ind_target_grp` (List[str]): List of target groups for individual modeling.
- `num_pop_folds` (int): Number of folds for cross-validation in population modeling.
- `num_ind_folds` (int): Number of folds for cross-validation in individual modeling.

### Returns:
int: Identifier value based on the specified column return type.

````python
def segregate_identifier(
    index: str,
    column_return_type: str,
    POPULATION_EPOCH_SIZE_LIST: List[int],
    PATIENT_EPOCH_SIZE_LIST: List[int],
    ind_target_grp: List[str],
    num_pop_folds: int,
    num_ind_folds: int,
) -> int:
    """
    Function to segregate identifiers based on the specified column return type.

    Args:
    - index (str): Identifier value.
    - column_return_type (str): Type of column to return.
    - POPULATION_EPOCH_SIZE_LIST (List[int]): List of epoch sizes for population modeling.
    - PATIENT_EPOCH_SIZE_LIST (List[int]): List of epoch sizes for patient modeling.
    - ind_target_grp (List[str]): List of target groups for individual modeling.
    - num_pop_folds (int): Number of folds for cross-validation in population modeling.
    - num_ind_folds (int): Number of folds for cross-validation in individual modeling.

    Returns:
    int: Identifier value based on the specified column return type.
    """
    index = str(index).upper()
    if column_return_type == "Pop_Epoch_Size":
        for i in POPULATION_EPOCH_SIZE_LIST:
            if str(i) + "_POP" in index:
                return i
    elif column_return_type == "Ind_Epoch_Size":
        for j in PATIENT_EPOCH_SIZE_LIST:
            if str(j) + "_IND" in index:
                return j
    elif column_return_type == "Patient_ID":
        for i in ind_target_grp:
            if str(i) in index:
                return i
    elif column_return_type == "Pop_Fold":
        for i in range(1, num_pop_folds + 1):
            if "POP_FOLD_" + str(i) in index:
                return i
    elif column_return_type == "Ind_Fold":
        for i in range(1, num_ind_folds + 1):
            if "IND_FOLD_" + str(i) in index:
                return i
````

## Function: `load_csv_to_df`

### Description:
This function loads a CSV file into a DataFrame and processes data according to specified columns. It maps each index value to the appropriate identifier based on the specified column return type.

### Arguments:
- `file_path` (str): Path to the CSV file.
- `data_processing_columns` (List[str]): List of columns for data processing.
- `POPULATION_EPOCH_SIZE_LIST` (List[int]): List of epoch sizes for population modeling.
- `PATIENT_EPOCH_SIZE_LIST` (List[int]): List of epoch sizes for patient modeling.
- `ind_target_grp` (List[str]): List of target groups for individual modeling.
- `num_pop_folds` (int): Number of folds for cross-validation in population modeling.
- `num_ind_folds` (int): Number of folds for cross-validation in individual modeling.

### Returns:
pd.DataFrame: Processed DataFrame.

````python
def load_csv_to_df(
    file_path: str,
    data_processing_columns: List[str],
    POPULATION_EPOCH_SIZE_LIST: List[int],
    PATIENT_EPOCH_SIZE_LIST: List[int],
    ind_target_grp: List[str],
    num_pop_folds: int,
    num_ind_folds: int,
) -> pd.DataFrame:
    """
    Load CSV file into a DataFrame and process data according to specified columns.

    Args:
    - file_path (str): Path to the CSV file.
    - data_processing_columns (List[str]): List of columns for data processing.
    - POPULATION_EPOCH_SIZE_LIST (List[int]): List of epoch sizes for population modeling.
    - PATIENT_EPOCH_SIZE_LIST (List[int]): List of epoch sizes for patient modeling.
    - ind_target_grp (List[str]): List of target groups for individual modeling.
    - num_pop_folds (int): Number of folds for cross-validation in population modeling.
    - num_ind_folds (int): Number of folds for cross-validation in individual modeling.

    Returns:
    pd.DataFrame: Processed DataFrame.
    """
    df = pd.read_csv(file_path, index_col=0)
    for i in data_processing_columns:
        df[i] = df.index.map(
            lambda x: segregate_identifier(
                x,
                i,
                POPULATION_EPOCH_SIZE_LIST,
                PATIENT_EPOCH_SIZE_LIST,
                ind_target_grp,
                num_pop_folds,
                num_ind_folds,
            )
        )

    return df
````

### Returns:
None

````python
def resave_eval_df(
    eval_data_type: str,
    RESULT_DIR: str,
    Architecture_GROUP: str,
    bool_remove_target: bool,
    data_processing_columns: List[str],
    POPULATION_EPOCH_SIZE_LIST: List[int],
    PATIENT_EPOCH_SIZE_LIST: List[int],
    ind_target_grp: List[str],
    num_pop_folds: int,
    num_ind_folds: int,
) -> None:
    """
    Resave evaluation DataFrame with sorted and processed data.

    Args:
    - eval_data_type (str): Type of evaluation data (individual or population).
    - RESULT_DIR (str): Directory path to save results.
    - Architecture_GROUP (str): Grouping identifier for architecture selection.
    - bool_remove_target (bool): Flag indicating whether to remove the target variable.
    - data_processing_columns (List[str]): List of columns for data processing.
    - POPULATION_EPOCH_SIZE_LIST (List[int]): List of epoch sizes for population modeling.
    - PATIENT_EPOCH_SIZE_LIST (List[int]): List of epoch sizes for patient modeling.
    - ind_target_grp (List[str]): List of target groups for individual modeling.
    - num_pop_folds (int): Number of folds for cross-validation in population modeling.
    - num_ind_folds (int): Number of folds for cross-validation in individual modeling.

    Returns:
    None
    """
    if bool_remove_target:
        suffix = "_evaluation_metrics_TargetRemoved.csv"
    else:
        suffix = "_evaluation_metrics.csv"

    if eval_data_type == "individual":
        eval_data_path = "\diagrams_n_individual_models\Ind_"
    elif eval_data_type == "population":
        eval_data_path = "\diagrams_n_population_models\Pop_"

    file_path = os.path.join(
        RESULT_DIR, Architecture_GROUP, eval_data_path, Architecture_GROUP + suffix
    )

    if os.path.isfile(file_path):
        temp_unsorted_eval_data_df = load_csv_to_df(
            file_path,
            data_processing_columns,
            POPULATION_EPOCH_SIZE_LIST,
            PATIENT_EPOCH_SIZE_LIST,
            ind_target_grp,
            num_pop_folds,
            num_ind_folds,
        )

        temp_sorted_eval_data_df = temp_unsorted_eval_data_df.sort_values(
            by=data_processing_columns, ascending=True
        ).drop(columns=data_processing_columns, errors="ignore")

        output_path = os.path.join(file_path)
        temp_sorted_eval_data_df.to_csv(output_path, mode="w")
````

## Code Execution: Resaving Evaluation DataFrames

### Description:
This code iterates over lists of evaluation data types and boolean flags for removing targets. For each combination of evaluation data type and remove target flag, it calls the `resave_eval_df` function with appropriate arguments to resave the evaluation DataFrames.

````python
data_processing_columns = [
    "Patient_ID",
    "Pop_Epoch_Size",
    "Pop_Fold",
    "Ind_Epoch_Size",
    "Ind_Fold",
]

bool_remove_target_list = [True, False]
eval_data_type_list = ["individual", "population"]
for i in eval_data_type_list:
    for j in bool_remove_target_list:
        resave_eval_df(
            i,
            RESULT_DIR,
            Architecture_GROUP,
            j,
            data_processing_columns,
            POPULATION_EPOCH_SIZE_LIST,
            PATIENT_EPOCH_SIZE_LIST,
            ind_target_grp,
            num_pop_folds,
            num_ind_folds,
        )

del data_processing_columns
del bool_remove_target_list
del eval_data_type_list
````
