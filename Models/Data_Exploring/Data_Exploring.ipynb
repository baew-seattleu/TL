# Importing Libraries


```python
import numpy as np
import os
import pandas as pd
from pathlib import Path

pd.set_option("display.max_columns", None)
```

```python
# SET DIRECTORY
execution_path = "../../"
# SETTING GLOBALS AND FILEPATHS

# Data to use:
source_quantile = "0.2"
source_file_path = f"{execution_path}Input_Data/Og_Data/labeled_normal_data/all_pids_q={source_quantile}.npy"

print("source_file_path", source_file_path)

# Target Groups
ind_target_grp = [
    "SB-001",
    "SB-003",
    "SB-008",
    "SB-011",
    "SB-012",
    "SB-022",
    "SB-023",
    "SB-025",
    "SB-028",
    "SB-037",
    "SB-043",
    "SB-056",
    "SB-059",
    "SB-060",
    "SB-070",
    "SB-071",
    "SB-073",
    "SB-078",
    "SB-079",
    "SB-080",
    "SB-082",
    "SB-083",
    "SB-112",
]
```

# Importing Data


```python
# CONVERT Data TO NUMPY ARRAYS
data_ndarr = np.load(source_file_path, allow_pickle=True)
print("data_ndarr type :", type(data_ndarr))

datadict = data_ndarr.item()
print("datadict type :", type(datadict))
```

```python
def data_groups(ind_target_grp, datadict, filename, execution_path):
    keys_list = datadict.keys()

    status_list = []
    patient_list = []


    total_rows_list = []
    class_1_list = []
    class_0_list = []


    for i in keys_list:
        if i in ind_target_grp:
            patient_status = "Used"

        else:
            patient_status = "Ignored"

        status_list.append(patient_status)
        patient_list.append(i)


        total_rows_list.append(datadict[i].shape[0])
        class_1_list.append(datadict[i]["class"].value_counts()[1.0])
        class_0_list.append(datadict[i]["class"].value_counts()[0.0])


    df = (
        pd.DataFrame(
            zip(status_list, patient_list, total_rows_list, class_1_list, class_0_list),
            columns=["status", "patient_id", "total_rows", "class_1", "class_0"],
        )

        .sort_values(
            by=["status", "patient_id", "total_rows", "class_1", "class_0"],
            ascending=[False, True, True, True, True],
        )
        .reset_index(drop=True)
    )

    output_path = os.path.join(f"{execution_path}Output_Results/Data_Exploring/")

    # Create directory for this patient
    if not (Path(output_path).exists() and Path(output_path).is_dir()):
        os.makedirs(output_path)

    df.to_csv(output_path + filename, mode="w")


filename = "explored_data.csv"
data_groups(ind_target_grp, datadict, filename, execution_path)
```
